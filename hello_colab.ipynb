# Hello from Colab

This notebook defines a simple addition function and prints a greeting and the sum of 2 and 3. It uses only Python's built-in features; no extra packages are needed.

```python
def add(a, b):
    return a + b


print("Hello from Colab!")
print(add(2, 3))


def multiply(a, b):
    return a * b


print(multiply(2, 3))
```
